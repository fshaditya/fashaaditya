Pengerjaan mengikuti urutan langkah pada modul: import CSV/DataFrame, `shape`, `describe`, `mode` dan `var`, `quantile`, `skew` dan `kurt`, lalu histogram.

## 1. Import data `.csv` dan membuat `DataFrame`

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

Password_XX = pd.read_csv('BeratProduk.csv')
Password_XX.head()
```

## 2. Mengetahui banyaknya data dengan script `shape`

```python
Password_XX.shape
```

## 3. Minimum, maksimum, mean, standar deviasi, kuartil 1, 2, dan 3 dengan `describe`

```python
print("\nStatistik Deskriptif:")
print(Password_XX.describe())
```

## 4. Modus dengan `mode` dan variansi dengan `var`

```python
# Mencari Modus
mode = Password_XX.mode()
print("Modus:", mode.iloc[0, 0])

# Menghitung Varians
variance = Password_XX.var()
print("Varians:", variance.iloc[0])
```

## 5. Mencari nilai persentil dengan `quantile`

```python
# Menghitung persentil ke-25
percentile25 = Password_XX.quantile(0.25)
print("Persentil ke-25:", percentile25.iloc[0])

# Menghitung persentil ke-50
percentile50 = Password_XX.quantile(0.50)
print("Persentil ke-50:", percentile50.iloc[0])

# Menghitung persentil ke-75
percentile75 = Password_XX.quantile(0.75)
print("Persentil ke-75:", percentile75.iloc[0])
```

## 6. Skewness dan kurtosis

```python
# Menghitung skewness
skewness = Password_XX.skew()
print("Skewness:", skewness.iloc[0])

# Menghitung kurtosis
kurt = Password_XX.kurt()
print("Kurtosis:", kurt.iloc[0])
```

## 7. Histogram

```python
# Menghitung jumlah kelas
n = len(Password_XX)
k = int(1+3.3*np.log10(n))
print("Jumlah Kelas (k):", k)

# Membuat histogram
plt.hist(Password_XX.iloc[:,0], bins=k, edgecolor='black')
plt.title('Histogram Berat Produk (gram)')
plt.xlabel('Nilai')
plt.ylabel('Frekuensi')
plt.show()
```
