```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
d1n2k38 = pd.read_csv('pop1n2.csv')
d1n2k38.head()
```

```python
#menghitung jumlah kelas
n = len(d1n2k38)
k = int(1+3.3*np.log10(n))
#menampilkan jumlah kelas
print("Jumlah Kelas (k)", k)
#membuat histogram
plt.hist(d1n2k38, bins=k, edgecolor='black')
#menambahkan judul dan label sumbu
plt.title ('Populasi 01 dengan n = 2')
plt.xlabel('Nilai')
plt.ylabel('Frekuensi')
#menampilkan histogram
plt.show()
```
